#### Indian startup funding dataset

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


df = pd.read_csv("india_startup_funding_sample.csv")
# Which 5 sectors raised the most total funding?

top_sector = df.groupby("sector")['amount_usd'].sum().sort_values(ascending=False).head(5)
amount = top_sector.to_numpy()
amounts_in_billions = amount / 1e9
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
plt.figure(figsize=(12, 5))
plt.bar(top_sector.index, amounts_in_billions, width=0.5, color=colors)
plt.ylabel('Amount (in Billions)')
plt.title('Total Amount by Category')
plt.show()

# print(top_sector.index)



```
