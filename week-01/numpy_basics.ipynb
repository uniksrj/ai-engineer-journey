### NumPy provides quick utility functions for creating arrays filled with zeros, ones, or ranges


```python
import numpy as np 

listOne = np.zeros((3,3))
print(listOne)

listTwo = np.ones((3,3))
print(listTwo)

listThree = np.arange(1,10,2)
print(listThree)
```

## numpy.fromiter() function – Python
### creates a one-dimensional array from an iterable object.

```python
import numpy as np

iterable = (x**3 for x in range(4))
myTxt = np.fromiter(iterable, int)
print(myTxt)
```

## numpy.arange() in Python
### numpy.arange() function creates an array of evenly spaced values within a given interval. It is similar to Python's built-in range() function but returns a NumPy array instead of a list.

```python
import numpy as np

list = np.arange(5,10)
print(list)
```

## NumPy - linspace() Function
### The numpy.linspace() function is used to generate an array of evenly spaced values between two specified numbers. Instead of defining a step size, the total number of required values is specified and NumPy automatically calculates the spacing between them.

#### Syntax
#### numpy.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)

#### Parameters:

##### start: Starting value (default 0)
##### stop: Ending value of the range
##### num: Number of values to generate (default 50)
##### endpoint: Includes stop if True (default True)
##### retstep: Returns step size if True (default False)
##### dtype: Output array data type
##### axis: Axis for generation when inputs are array-like (default 0)

```python
import numpy as np

a = np.linspace(0,1,10,endpoint=False)
print(a)
```
