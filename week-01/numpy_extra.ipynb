```python
import numpy as np

rng = np.random.default_rng(42)

data = rng.normal(size=1000)
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))

data = rng.random((5,2))
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))

data = rng.integers(0,10,size=6)
print(data)

t = rng.shuffle(data, axis=0)
print(t)

t = rng.permutation(data, axis=0)
print(t)
```

```python
import numpy as np

# This is practice set of details


def analyze_scores(scores: np.ndarray) -> dict[str, float | np.ndarray]:
    ret_data = {}
    # Return total, mean, standard deviation,
    total = np.sum(scores)
    ret_data['sum'] = total
    print("total :",total)
    mean = np.mean(scores)
    ret_data['mean'] = round(mean,2)
    print("Mean :",mean)
    standard_deviation = np.std(scores)
    ret_data['std'] = round(standard_deviation,2)
    print("Standard Deviation :",standard_deviation)

    # highest score, lowest score,
    maxNum = np.max(scores)
    ret_data['max'] = maxNum
    print("Max :",maxNum)
    minNumber = np.min(scores)
    ret_data['min'] = minNumber
    print("Min :",minNumber)

    # and scores above the average
    averageCount = np.average(scores)
    ret_data['average'] = averageCount
    print("Average :",averageCount)   
    
    return ret_data

scores = np.array([55, 72, 89, 91, 64, 78, 95])
result = analyze_scores(scores)
print(result)
```
