```python
import pandas as pd
import numpy as np

data = pd.read_csv("employe.csv")
df = pd.DataFrame(data, columns=["employee", "department", "salary", "experience_years", "city"])
dict_data = df.to_dict('records')
for record in dict_data:
    record['annual_salary'] = record['salary'] * 12
    record['salary_category'] = ("High" if record['salary'] > 75000 else "Low" if record['salary'] < 60000 else "Medium")
    record['senior'] = True if record['experience_years'] >= 5 else False

df = pd.DataFrame(dict_data, columns=["employee", "department", "salary", "experience_years", "city","annual_salary","salary_category","senior"])
df
    
```

```python
import pandas as pd
import numpy as np

data = pd.read_csv("employe.csv")
df = pd.DataFrame(data, columns=["employee", "department", "salary", "experience_years", "city"])
numpyData = df.to_numpy()
# print(numpyData)

salary = numpyData[:, 2]
experience_years = numpyData[:, 3]

annual_salary = salary * 12
salary_category = np.where(salary > 75000, "High", np.where(salary < 60000, "Low", "Medium"))
senior = np.where(experience_years >= 5, True, False)

new_numpyData = np.column_stack((numpyData, annual_salary, salary_category, senior))
df = pd.DataFrame(new_numpyData, columns=["employee", "department", "salary", "experience_years", "city","annual_salary","salary_category","senior"])
df
# print(new_numpyData)

# for data in numpyData:
#     annual_salary = data[2] * 12
#     np.append((data, annual_salary))  # Adding annual salary to each row
#     np.append(data, "High" if data[2] > 75000 else "Low" if data[2] < 60000 else "Medium")  # Adding salary category
#     np.append(data, True if data[3] >= 5 else False)  # Adding seniority status

# print(numpyData)
# df = pd.DataFrame(numpyData, columns=["employee", "department", "salary", "experience_years", "city", "annual_salary", "salary_category", "senior"])
# df   
```
