## Add and modify columns

```python
import pandas as pd
import numpy as np

data = pd.read_csv("employe.csv")
df = pd.DataFrame(data, columns=["employee", "department", "salary", "experience_years", "city"])
dict_data = df.to_dict('records')
for record in dict_data:
    record['annual_salary'] = record['salary'] * 12
    record['salary_category'] = ("High" if record['salary'] > 75000 else "Low" if record['salary'] < 60000 else "Medium")
    record['senior'] = True if record['experience_years'] >= 5 else False

df = pd.DataFrame(dict_data, columns=["employee", "department", "salary", "experience_years", "city","annual_salary","salary_category","senior"])
df
    
```

## Missing values

```python
import pandas as pd
import numpy as np

df = pd.read_csv("employe.csv")
df_missing = df.copy()
df_missing.loc[2, "salary"] = np.nan
df_missing.loc[4,"city"] = np.nan

average_salery = df_missing['salary'].mean().round(2)
missing_rows = df_missing[df_missing.isna().any(axis=1)]
missing_rows['salary'] = missing_rows['salary'].fillna(average_salery)
missing_rows['city'] = missing_rows['city'].fillna("Unknown")
missing_rows
```

## GroupBy

```python
import pandas as pd
import numpy as np


df = pd.read_csv("employe.csv")

df.groupby("department").mean(numeric_only=True)
df.groupby("department").sum(numeric_only=True)
df.groupby("city")["experience_years"].mean().round(2)

df.groupby("department").max(numeric_only=True)
```

## Sorting and exporting

```python
import pandas as pd
import numpy as np


df = pd.read_csv("employe.csv")

df.sort_values(by="salary", ascending=False)
df.sort_values(by=["department", "salary"], ascending=False)
new_data = df.sort_values(by=["salary"], ascending=False).head(3)

new_data.to_csv("employees_cleaned.csv", index=False)
```

## Create two charts

#### Bar chart: average salary by department



```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


df = pd.read_csv("employe.csv")
grouped = df.groupby("department")["salary"].mean().round(2)

x = grouped.index.to_numpy()
y = grouped.to_numpy()
plt.bar(x,y,color=["red","blue","purple","orange","green"])
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
# plt.show()
plt.savefig("output/average_salary_by_department.png")
```

#### Histogram: distribution of employee salaries

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


df = pd.read_csv("employe.csv")

plt.hist(df['salary'], bins=10, color="blue", edgecolor="black")
plt.title("Distribution of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
# plt.show()
plt.savefig("output/distribution _salary_employee .png")
```
